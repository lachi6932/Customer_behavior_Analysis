```python
import pandas as pd
df=pd.read_csv('customer_shopping_behavior.csv');
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df['Review Rating'].head(30)
```

```python
df.columns = df.columns.str.replace('_', '', regex=False)
```

```python
df.columns
```

```python
df.columns = df.columns.str.strip().str.lower()
df.columns = df.columns.str.replace(' ', '_', regex=False)
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
print(df.columns.tolist())
```

```python
#create columns age_group
labels=['Young Adult','Adult','middle_aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create colimn purchase_frequence_days
frequence_mapping = {
    'Weekly': 7,
    'Fortnightly': 14,
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Every 3 Months': 90,
    'Annually': 365
}
```

```python
df['purchase_frequence_days'] = df['frequency_of_purchases'].map(frequence_mapping)
```

```python
print(df['purchase_frequence_days'].isna().sum())
```

```python
df['purchase_frequence_days'] = df['purchase_frequence_days'].astype(int)
```

```python
df[['purchase_frequence_days', 'frequency_of_purchases']].head(10)
```

```python
import psycopg

conn = psycopg.connect(
    host="localhost",
    port=5432,
    dbname="postgres",
    user="postgres",
    password="root"
)

conn.autocommit = True

print("Connected to PostgreSQL!")
```

```python
cur = conn.cursor()

cur.execute("CREATE DATABASE customer_behavior")

print("Database created successfully!")
```

```python
cur.close()
conn.close()
```

```python
from sqlalchemy import create_engine

username = "postgres"
password = "root"
host = "localhost"
port = "5432"
database = "customer_behavior"

engine = create_engine(
    f"postgresql+psycopg://{username}:{password}@{host}:{port}/{database}"
)

table_name = "customer"

df.to_sql(table_name, engine, if_exists="replace", index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
